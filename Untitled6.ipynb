```python
import pandas as pd
import matplotlib.pyplot as plt

# Load all files with the correct encoding using relative paths
df_batting = pd.read_csv("BATTING.csv", encoding='latin1')
df_bowling = pd.read_csv("BOWLING.csv", encoding='latin1')
df_matches = pd.read_csv("Cricket_WC23.csv", encoding='latin1')
df_players = pd.read_csv("PLAYERS_INFO.csv", encoding='latin1')

# Convert RUNS column to numeric, forcing any invalid entries to NaN
df_batting['RUNS'] = pd.to_numeric(df_batting['RUNS'], errors='coerce')

# Group by player and sum their runs, then sort
top_scorers = df_batting.groupby("BATTING")["RUNS"].sum().sort_values(ascending=False).head(10)

print("\nTop 10 Run Scorers (Fixed):")
print(top_scorers)

# Plot the numeric data
top_scorers.plot(kind="bar", figsize=(10, 5), color="steelblue")
plt.title("Top 10 Run Scorers - 2023 World Cup")
plt.xlabel("Player")
plt.ylabel("Total Runs")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Clean up column names by removing extra spaces
df_batting.columns = df_batting.columns.str.strip()

# Convert 'SIXES' and 'FOURS' columns to numeric to calculate sums
df_batting['SIXES'] = pd.to_numeric(df_batting['SIXES'], errors='coerce')
df_batting['FOURS'] = pd.to_numeric(df_batting['FOURS'], errors='coerce')

sixes = df_batting.groupby("BATTING")["SIXES"].sum().sort_values(ascending=False).head(10)
fours = df_batting.groupby("BATTING")["FOURS"].sum().sort_values(ascending=False).head(10)

print("\nTop 10 Six Hitters:")
print(sixes)
print("\nTop 10 Four Hitters:")
print(fours)

# Ensure BALLS column is numeric
df_batting['BALLS'] = pd.to_numeric(df_batting['BALLS'], errors='coerce')

# Calculate Best Strike Rates using correct dataset and columns
agg = df_batting.groupby("BATTING").agg(runs=("RUNS", "sum"), balls=("BALLS", "sum"))
agg["strike_rate"] = (agg["runs"] / agg["balls"]) * 100

# Filter for players who faced at least 100 balls to avoid skewed data from tailenders
best_sr = agg[agg["balls"] >= 100].sort_values("strike_rate", ascending=False).head(10)

print("\nTop 10 Best Strike Rates (Min 100 Balls Faced):")
print(best_sr)

# Plot Best Strike Rates
data = best_sr["strike_rate"]
colors_sr = plt.cm.tab10(range(len(data)))
data.plot(kind="barh", figsize=(10, 5), color=colors_sr)
plt.title("Best Strike Rates (min 100 balls) - 2023 World Cup")
plt.xlabel("Strike Rate")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Plot Most Sixes
colors_sixes = plt.cm.tab10(range(len(sixes)))
sixes.plot(kind="barh", figsize=(10, 5), color=colors_sixes)
plt.title("Most Sixes - 2023 World Cup")
plt.xlabel("Sixes")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Clean player name spaces in both dataframes before merging
df_batting['BATTING'] = df_batting['BATTING'].str.strip()
df_players['Player'] = df_players['Player'].str.strip()

# Merge batting data with players info to get the team name for each player
df_merged = pd.merge(df_batting, df_players, left_on="BATTING", right_on="Player", how="left")

# Group by 'Team' and sum 'RUNS'
team_runs = df_merged.groupby("Team")["RUNS"].sum().sort_values(ascending=False)
print("\nTotal Runs by Team:")
print(team_runs)

# Plot Total Runs by Team
colors = plt.cm.tab10(range(len(team_runs)))
team_runs.plot(kind="bar", figsize=(10, 5), color=colors)
plt.title("Total Runs by Team - 2023 World Cup")
plt.xlabel("Team")
plt.ylabel("Total Runs")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Calculate player consistency correctly using df_batting and correct column labels
consistency = df_batting.groupby("BATTING")["RUNS"].agg(["sum", "count", "mean"])
consistency.columns = ["total_runs", "matches", "avg_runs"]

# Only players with at least 5 innings, for a fair comparison
top_avg = consistency[consistency["matches"] >= 5].sort_values("avg_runs", ascending=False).head(10)
print("\nBest Batting Average per Innings (Min 5 innings):")
print(top_avg)

# Plot batting averages
colors_avg = plt.cm.tab10(range(len(top_avg)))
top_avg["avg_runs"].plot(kind="barh", figsize=(10, 5), color=colors_avg)
plt.title("Best Batting Average per Innings (min 5 innings) - 2023 World Cup")
plt.xlabel("Average Runs")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Aggregation using correct uppercase headers
agg_scatter = df_batting.groupby("BATTING").agg(runs=("RUNS", "sum"), balls=("BALLS", "sum"))
agg_scatter = agg_scatter[agg_scatter["balls"] >= 50]   # drop players with very few balls

plt.figure(figsize=(10, 6))
plt.scatter(agg_scatter["balls"], agg_scatter["runs"], s=80, alpha=0.7, color="teal")

# Label the top 5 run scorers
for player in agg_scatter.sort_values("runs", ascending=False).head(5).index:
    plt.annotate(player, (agg_scatter.loc[player, "balls"], agg_scatter.loc[player, "runs"]),
                 xytext=(5, 5), textcoords="offset points")

plt.title("Runs vs. Balls Faced - 2023 World Cup")
plt.xlabel("Balls Faced")
plt.ylabel("Total Runs")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Correct summary calculations using uppercase df_batting metrics
top_scorer = df_batting.groupby("BATTING")["RUNS"].sum().idxmax()
top_scorer_runs = df_batting.groupby("BATTING")["RUNS"].sum().max()

agg_summary = df_batting.groupby("BATTING").agg(runs=("RUNS", "sum"), balls=("BALLS", "sum"))
agg_summary["strike_rate"] = agg_summary["runs"] / agg_summary["balls"] * 100
fastest = agg_summary[agg_summary["balls"] >= 100]["strike_rate"].idxmax()
fastest_sr = agg_summary[agg_summary["balls"] >= 100]["strike_rate"].max()

most_sixes = df_batting.groupby("BATTING")["SIXES"].sum().idxmax()
most_sixes_n = df_batting.groupby("BATTING")["SIXES"].sum().max()

print(f"\nTop run scorer: {top_scorer} ({top_scorer_runs} runs)")
print(f"Fastest scorer (min 100 balls): {fastest} (SR {fastest_sr:.1f})")
print(f"Most sixes: {most_sixes} ({most_sixes_n})")
```
